## 1. 🕹️ Control the Robot in Real Time

Open the **terminal** and start typing commands directly to your robot!

````python
ls  # list files
````

````python
cd ~/ros2_ws/src/  # Move to the /home/user/ros2_ws/src directory
````

````python
ros2 topic list # List ROS topics
````

````python
ros2 topic pub /hello std_msgs/msg/String "{data: 'Hi from the rosject'}" --rate 1  # Publish a message in a ROS2 Topic named /hello
````

````python
ros2 topic echo /hello  # "Listen" to the /hello topic, in another terminal, after running "ros2 topic pub".
````

````python
ros2 topic pub /cmd_vel geometry_msgs/msg/Twist '{linear:  {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.1}}' # Rotating a robot.
````

Send movement commands, launch nodes, run scripts — **you're in full control**.
